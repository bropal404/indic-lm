````python
import math
import os
import time
import random
import json
from pathlib import Path

import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from tqdm.auto import tqdm
import wandb

from huggingface_hub import HfApi, hf_hub_download
````

````python
# TOKENIZER CODE 

import argparse
import os
import re
import unicodedata
import random
import pickle
import json
import heapq
import time
from pathlib import Path
from collections import defaultdict, Counter
from tqdm import tqdm 

random.seed(69)
BPE_MERGES = 20000 # Default


def clean_text(text, language="non-english"):
    text = unicodedata.normalize('NFKC', text)
    text = "".join(ch for ch in text if unicodedata.category(ch)[0] != "C" or ch in ["\n", "\t"])
    text = re.sub(r'\s+', ' ', text)
    if language.lower() == "english":
        text = text.encode("ascii", "ignore").decode("ascii")
    return text.strip()


def load_data(file_path, text_field="text", limit=None):
    path_name = os.path.basename(file_path).lower()
    language = "english" if ("english" in path_name or path_name.startswith("cc100_en") or path_name.endswith("_en.jsonl")) else "non-english"
    data = []
    with open(file_path, 'r', encoding='utf-8') as f:
        if file_path.endswith('.jsonl'):
            for line in f:
                if not line.strip():
                    continue
                try:
                    obj = json.loads(line)
                    val = obj.get(text_field) or obj.get('text')
                    if val:
                        cleaned = clean_text(str(val), language)
                        if cleaned:
                            data.append(cleaned)
                            if limit and len(data) >= limit:
                                break
                except json.JSONDecodeError:
                    continue
        else:
            text = f.read()
            cleaned = clean_text(text, language)
            if cleaned:
                data.append(cleaned)
    return data


def split_data(data):
    data = list(data)
    random.shuffle(data)
    n = len(data)
    train_end = int(n * 0.5)
    val_end = int(n * 0.75)
    return data[:train_end], data[train_end:val_end], data[val_end:]


# --- Tokenizers ---

class Tokenizer:
    def train(self, data):
        pass
    
    def tokenize(self, text):
        raise NotImplementedError
    
    def save(self, path):
        pass
    
    def load(self, path):
        pass


class WhitespaceTokenizer(Tokenizer):
    def tokenize(self, text):
        return [t for t in re.split(r'(\s+)', text) if t and not t.isspace()]


class RegexTokenizer(Tokenizer):
    def __init__(self):
        self.pattern = re.compile(r"[\w\u0900-\u097F]+|[^\w\s]", re.UNICODE)
    
    def tokenize(self, text):
        return self.pattern.findall(text)


class customBPE(Tokenizer):
    def __init__(self, num_merges=BPE_MERGES, min_pair_frequency=2):
        self.num_merges = num_merges
        self.min_pair_frequency = min_pair_frequency
        self.merges = {}  # (left, right) -> merged string
        self.vocab = {}
        self._token_cache = {}
        self.pre_tokenizer = RegexTokenizer()

    def _build_word_frequencies(self, data):
        token_frequencies = Counter()
        for line in tqdm(data, desc="Pre-tokenizing words"):
            words = self.pre_tokenizer.tokenize(line)
            for word in words:
                token_frequencies[tuple(list(word) + ["</w>"])] += 1
        return token_frequencies

    @staticmethod
    def _get_pairs_with_positions(sequence):
        """Returns dict of pair -> list of index positions in sequence"""
        pairs = defaultdict(list)
        for i in range(len(sequence) - 1):
            pairs[(sequence[i], sequence[i + 1])].append(i)
        return pairs

    @staticmethod
    def _merge_sequence_fast(sequence, pair):
        left, right = pair
        merged = left + right
        new_seq = []
        i = 0
        n = len(sequence)
        while i < n:
            if i < n - 1 and sequence[i] == left and sequence[i + 1] == right:
                new_seq.append(merged)
                i += 2
            else:
                new_seq.append(sequence[i])
                i += 1
        return tuple(new_seq)

    def _build_vocab(self, sequences):
        # Reserve special tokens up front: <unk> for OOV pieces, <S>/</S>/<pad> as
        # sentinel tokens the training loop expects, and </w> as an explicit
        # word-boundary marker (needed so decode() can reinsert spaces).
        tokens = {"<unk>": 0, "<S>": 1, "</S>": 2, "<pad>": 3, "</w>": 4}
        for sequence in sequences:
            for token in sequence:
                if token == "</w>":
                    continue
                clean_token = token.replace("</w>", "")
                if clean_token and clean_token not in tokens:
                    tokens[clean_token] = len(tokens)
        return tokens

    def train(self, data):
        print("Training Optimized BPE Tokenizer...")
        t0 = time.time()
        word_frequencies = self._build_word_frequencies(data)

        # Sequences and their occurrence frequencies
        sequences = list(word_frequencies.keys())
        frequencies = [word_frequencies[sequence] for sequence in sequences]
        print(f"Unique words / initial sequence types: {len(sequences):,}")

        # Index pair counts and inverted index (pair -> set of word_ids)
        pair_counts = Counter()
        pair_to_words = defaultdict(set)

        for word_id, sequence in enumerate(sequences):
            freq = frequencies[word_id]
            for i in range(len(sequence) - 1):
                p = (sequence[i], sequence[i + 1])
                pair_counts[p] += freq
                pair_to_words[p].add(word_id)

        # Max-heap for constant time O(1) top pair lookup
        heap = [(-count, pair) for pair, count in pair_counts.items() if count >= self.min_pair_frequency]
        heapq.heapify(heap)

        print(f"Starting up to {self.num_merges:,} merges...")
        progress = tqdm(total=self.num_merges, desc="Merging pairs")

        while len(self.merges) < self.num_merges and heap:
            neg_count, best_pair = heapq.heappop(heap)
            count = -neg_count
            current_count = pair_counts.get(best_pair, 0)

            # Lazy heap invalidation check
            if count != current_count:
                continue
            if current_count < self.min_pair_frequency:
                break

            self.merges[best_pair] = best_pair[0] + best_pair[1]
            affected_word_ids = pair_to_words.pop(best_pair, set())
            del pair_counts[best_pair]

            left, right = best_pair

            for word_id in affected_word_ids:
                old_sequence = sequences[word_id]
                freq = frequencies[word_id]

                # Fast check if pair still exists in word
                if left not in old_sequence:
                    continue

                # Remove old pairs
                for i in range(len(old_sequence) - 1):
                    p = (old_sequence[i], old_sequence[i + 1])
                    pair_counts[p] -= freq
                    if pair_counts[p] <= 0:
                        del pair_counts[p]
                        pair_to_words[p].discard(word_id)

                # Merge sequence
                new_sequence = self._merge_sequence_fast(old_sequence, best_pair)
                sequences[word_id] = new_sequence

                # Add new pairs
                for i in range(len(new_sequence) - 1):
                    p = (new_sequence[i], new_sequence[i + 1])
                    pair_counts[p] += freq
                    pair_to_words[p].add(word_id)
                    new_c = pair_counts[p]
                    if new_c >= self.min_pair_frequency:
                        heapq.heappush(heap, (-new_c, p))

            progress.update(1)

        progress.close()
        self.merge_ranks = {pair: idx for idx, pair in enumerate(self.merges.keys())}
        self.vocab = self._build_vocab(sequences)
        self._token_cache.clear()
        print(f"BPE Training Complete in {time.time() - t0:.2f}s! Learned {len(self.merges):,} merges, Vocab: {len(self.vocab):,} tokens.")

    def _bpe_word(self, word):
        """Apply learned merges to a single pre-tokenized word. Returns the
        subword pieces for that word (word-boundary marker stripped, since
        encode()/decode() track boundaries separately)."""
        cached = self._token_cache.get(word)
        if cached is not None:
            return cached

        if not hasattr(self, 'merge_ranks'):
            self.merge_ranks = {pair: idx for idx, pair in enumerate(self.merges.keys())}

        tokens = list(word) + ["</w>"]

        while len(tokens) > 1:
            pairs = [(tokens[i], tokens[i + 1]) for i in range(len(tokens) - 1)]
            best_pair = None
            earliest_rank = float('inf')

            for pair in pairs:
                rank = self.merge_ranks.get(pair)
                if rank is not None and rank < earliest_rank:
                    earliest_rank = rank
                    best_pair = pair

            if best_pair is None:
                break

            left, right = best_pair
            merged_tokens = []
            i = 0
            n = len(tokens)
            while i < n:
                if i < n - 1 and tokens[i] == left and tokens[i + 1] == right:
                    merged_tokens.append(left + right)
                    i += 2
                else:
                    merged_tokens.append(tokens[i])
                    i += 1

            tokens = merged_tokens

        cleaned_tokens = [t.replace('</w>', '') for t in tokens if t != '</w>']
        cleaned_tokens = [token for token in cleaned_tokens if token]
        self._token_cache[word] = cleaned_tokens
        return cleaned_tokens

    def tokenize(self, text):
        words = self.pre_tokenizer.tokenize(text)
        result = []
        for word in words:
            result.extend(self._bpe_word(word))
        return result

    @property
    def vocab_size(self):
        return len(self.vocab)

    def encode(self, text):
        """Text -> list of token ids. Sentinel tokens ("<S>", "</S>", "<pad>",
        "<unk>") are passed straight through as a single id when the whole
        input matches one of them; otherwise the text is BPE-tokenized word
        by word, with an explicit "</w>" id inserted after each word so
        decode() can reconstruct spacing."""
        if text in ("<S>", "</S>", "<pad>", "<unk>") and text in self.vocab:
            return [self.vocab[text]]

        unk_id = self.vocab.get("<unk>", 0)
        boundary_id = self.vocab.get("</w>", unk_id)
        words = self.pre_tokenizer.tokenize(text)
        ids = []
        for word in words:
            for tok in self._bpe_word(word):
                ids.append(self.vocab.get(tok, unk_id))
            ids.append(boundary_id)
        return ids

    def decode(self, ids):
        """List of token ids -> text. "</w>" ids become spaces; everything
        else is concatenated directly (subwords within a word have no
        separator)."""
        if not hasattr(self, '_id_to_token') or len(self._id_to_token) != len(self.vocab):
            self._id_to_token = {i: t for t, i in self.vocab.items()}

        pieces = []
        for i in ids:
            tok = self._id_to_token.get(i, "<unk>")
            if tok == "</w>":
                pieces.append(" ")
            elif tok in ("<S>", "</S>", "<pad>"):
                continue
            else:
                pieces.append(tok)
        return "".join(pieces).strip()

    def save(self, path):
        model = {
            "num_merges": self.num_merges,
            "min_pair_frequency": self.min_pair_frequency,
            "merges": self.merges,
            "vocab": self.vocab,
        }
        with open(path, 'wb') as f:
            pickle.dump(model, f)

    def load(self, path):
        with open(path, 'rb') as f:
            model = pickle.load(f)
        if isinstance(model, dict) and "merges" in model:
            self.num_merges = model.get("num_merges", self.num_merges)
            self.min_pair_frequency = model.get("min_pair_frequency", self.min_pair_frequency)
            self.merges = model["merges"]
            self.vocab = model.get("vocab", {})
        else:
            self.merges = model
            self.vocab = {}
        self.merge_ranks = {pair: idx for idx, pair in enumerate(self.merges.keys())}
        self._token_cache.clear()

    def save_artifacts(self, output_dir, name):
        output_dir = Path(output_dir)
        output_dir.mkdir(parents=True, exist_ok=True)

        model_path = output_dir / f"{name}.pkl"
        vocab_path = output_dir / f"{name}_vocab.json"
        merges_path = output_dir / f"{name}_merges.txt"
        config_path = output_dir / f"{name}_config.json"

        self.save(model_path)

        with open(vocab_path, "w", encoding="utf-8") as f:
            json.dump(self.vocab, f, ensure_ascii=False, indent=2)

        with open(merges_path, "w", encoding="utf-8") as f:
            for left, right in self.merges:
                f.write(f"{left} {right}\n")

        config = {
            "type": "custom_bpe",
            "num_merges": self.num_merges,
            "learned_merges": len(self.merges),
            "min_pair_frequency": self.min_pair_frequency,
            "vocab_size": len(self.vocab),
            "model_file": model_path.name,
            "vocab_file": vocab_path.name,
            "merges_file": merges_path.name,
        }
        with open(config_path, "w", encoding="utf-8") as f:
            json.dump(config, f, ensure_ascii=False, indent=2)

        return {
            "model": model_path,
            "vocab": vocab_path,
            "merges": merges_path,
            "config": config_path,
        }
````

## 6. Model

````python
class MultiHeadCausalAttention(nn.Module):
    def __init__(self, d_model, n_heads, dropout=0.1):
        super().__init__()
        assert d_model % n_heads == 0

        self.d_model = d_model
        self.n_heads = n_heads
        self.head_dim = d_model // n_heads

        self.q_proj = nn.Linear(d_model, d_model, bias=False)
        self.k_proj = nn.Linear(d_model, d_model, bias=False)
        self.v_proj = nn.Linear(d_model, d_model, bias=False)
        self.o_proj = nn.Linear(d_model, d_model, bias=False)

        self.attn_dropout = nn.Dropout(dropout)
        self.resid_dropout = nn.Dropout(dropout)

    def forward(self, x, return_attention=False):
        # x: (B, T, d_model)
        B, T, C = x.shape

        Q = self.q_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        K = self.k_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        V = self.v_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)

        if return_attention:
            # Fallback to manual attention if attention weights are explicitly requested
            scores = (Q @ K.transpose(-2, -1)) / math.sqrt(self.head_dim)
            causal_mask = torch.triu(
                torch.ones(T, T, device=x.device, dtype=torch.bool),
                diagonal=1
            )
            scores = scores.masked_fill(causal_mask, float('-inf'))
            attention = torch.softmax(scores, dim=-1)
            attention = self.attn_dropout(attention)
            out = attention @ V
        else:
            # Extremely fast FlashAttention (SDPA)
            out = torch.nn.functional.scaled_dot_product_attention(
                Q, K, V, 
                is_causal=True, 
                dropout_p=self.attn_dropout.p if self.training else 0.0
            )
            attention = None

        # (B,H,T,D) -> (B,T,C)
        out = out.transpose(1, 2).contiguous().view(B, T, C)
        out = self.o_proj(out)
        out = self.resid_dropout(out)

        if return_attention:
            return out, attention
        return out


class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff, dropout=0.1):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(d_model, d_ff),
            nn.GELU(),
            nn.Linear(d_ff, d_model),
            nn.Dropout(dropout)
        )

    def forward(self, x):
        return self.net(x)


class TransformerBlock(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, dropout=0.1):
        super().__init__()
        self.ln1 = nn.LayerNorm(d_model)
        self.attn = MultiHeadCausalAttention(d_model, n_heads, dropout)

        self.ln2 = nn.LayerNorm(d_model)
        self.ff = FeedForward(d_model, d_ff, dropout)

    def forward(self, x, return_attention=False):
        if return_attention:
            attn_out, attention = self.attn(
                self.ln1(x),
                return_attention=True
            )
        else:
            attn_out = self.attn(self.ln1(x))

        x = x + attn_out
        x = x + self.ff(self.ln2(x))

        if return_attention:
            return x, attention
        return x


class TransformerLM(nn.Module):
    def __init__(
        self,
        vocab_size,
        d_model=D_MODEL,
        n_layers=N_LAYERS,
        n_heads=N_HEADS,
        d_ff=D_FF,
        context_length=SEQ_LEN,
        dropout=DROPOUT
    ):
        super().__init__()

        self.d_model = d_model
        self.context_length = context_length

        self.token_embedding = nn.Embedding(vocab_size, d_model)
        self.position_embedding = nn.Embedding(context_length, d_model)
        self.embedding_dropout = nn.Dropout(dropout)

        self.blocks = nn.ModuleList([
            TransformerBlock(
                d_model=d_model,
                n_heads=n_heads,
                d_ff=d_ff,
                dropout=dropout
            )
            for _ in range(n_layers)
        ])

        self.ln_f = nn.LayerNorm(d_model)

        # Weight tying: output matrix shares token embedding weights.
        self.head = nn.Linear(d_model, vocab_size, bias=False)
        self.head.weight = self.token_embedding.weight

        self.apply(self._init_weights)

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            # Do not reinitialize the tied embedding/output matrix here.
            if module.weight is not self.token_embedding.weight:
                nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:
                nn.init.zeros_(module.bias)

        elif isinstance(module, nn.Embedding):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)

    def forward(self, input_ids, targets=None, return_attention=False):
        B, T = input_ids.shape
        assert T <= self.context_length

        # (B,T) -> (B,T,d_model)
        tok_emb = self.token_embedding(input_ids)

        # (T,) -> (T,d_model), broadcast over batch
        positions = torch.arange(T, device=input_ids.device)
        pos_emb = self.position_embedding(positions)

        x = self.embedding_dropout(tok_emb + pos_emb)

        attentions = []
        for block in self.blocks:
            if return_attention:
                x, attention = block(x, return_attention=True)
                attentions.append(attention)
            else:
                x = block(x)

        x = self.ln_f(x)

        # (B,T,d_model) -> (B,T,vocab_size)
        logits = self.head(x)

        loss = None
        if targets is not None:
            # Output logits already match targets because inputs are shifted ahead during data loading
            loss = nn.functional.cross_entropy(
                logits.view(-1, logits.size(-1)),
                targets.view(-1),
                ignore_index=-1
            )

        if return_attention:
            return logits, loss, attentions
        return logits, loss

    @torch.no_grad()
    def generate(
        self,
        input_ids,
        max_new_tokens=50,
        temperature=1.0,
        top_k=0,
        greedy=False
    ):
        self.eval()

        for _ in range(max_new_tokens):
            idx_cond = input_ids[:, -self.context_length:]
            logits, _ = self(idx_cond)
            logits = logits[:, -1, :]

            if greedy:
                idx_next = torch.argmax(logits, dim=-1, keepdim=True)
            else:
                if temperature <= 0:
                    raise ValueError("temperature must be > 0 unless greedy=True")

                logits = logits / temperature

                if top_k > 0:
                    values, _ = torch.topk(
                        logits,
                        min(top_k, logits.size(-1))
                    )
                    threshold = values[:, [-1]]
                    logits = logits.masked_fill(
                        logits < threshold,
                        -float("inf")
                    )

                probs = torch.softmax(logits, dim=-1)
                idx_next = torch.multinomial(probs, num_samples=1)

            input_ids = torch.cat([input_ids, idx_next], dim=1)

        return input_ids
````

## 9. Initialize Weights & Biases

````python
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)


def compute_perplexity(model, dataloader, device, max_batches=200):
    model.eval()
    total_nll = 0.0
    total_tokens = 0

    with torch.no_grad():
        for i, (x, y) in enumerate(dataloader):
            if i >= max_batches:
                break

            x, y = x.to(device), y.to(device)
            _, loss = model(x, y)

            n_tokens = y.numel()
            total_nll += loss.item() * n_tokens
            total_tokens += n_tokens

    if total_tokens == 0:
        return float("inf")

    avg_loss = total_nll / total_tokens
    return math.exp(avg_loss)


def verify_causal_mask(model, x, vocab_size):
    """Changing a future token must not change logits at an earlier position."""
    model.eval()

    x1 = x.clone()
    x2 = x.clone()

    t = x.shape[1] // 2
    x2[:, t + 1] = (x2[:, t + 1] + 1) % vocab_size

    with torch.no_grad():
        logits1, _ = model(x1)
        logits2, _ = model(x2)

    difference = (
        logits1[:, t, :] - logits2[:, t, :]
    ).abs().max().item()

    print(f"Maximum future-token influence at position {t}: {difference:.10e}")

    if difference < 1e-6:
        print("Causal masking verified.")
    else:
        raise AssertionError("Causal masking check failed.")


def attention_entropy(attention):
    # attention: (B,H,T,T)
    eps = 1e-10
    entropy = -(
        attention * torch.log(attention + eps)
    ).sum(dim=-1)

    # (H,)
    return entropy.mean(dim=(0, 2))


def mean_attention_distance(attention):
    # attention: (B,H,T,T)
    B, H, T, _ = attention.shape

    positions = torch.arange(
        T,
        device=attention.device,
        dtype=torch.float32
    )

    distances = (
        positions[None, :] - positions[:, None]
    ).abs()

    weighted = attention * distances

    # (B,H,T) -> (H,)
    return weighted.sum(dim=-1).mean(dim=(0, 2))


def load_tokenizer_from_hf(repo_id, token):
    api = HfApi(token=token)
    tokenizer_filename = None

    files = api.list_repo_files(repo_id=repo_id, token=token)
    pkl_files = [f for f in files if f.endswith(".pkl")]

    if pkl_files:
        for pf in pkl_files:
            if "20000" in pf or "artifacts_20000" in pf:
                tokenizer_filename = pf
                break
        if tokenizer_filename is None:
            tokenizer_filename = pkl_files[0]

    if tokenizer_filename is None:
        raise FileNotFoundError(f"No 20k tokenizer .pkl found in {repo_id}.")

    print(f"Downloading tokenizer from {repo_id}:{tokenizer_filename}...")
    local_path = hf_hub_download(
        repo_id=repo_id,
        filename=tokenizer_filename,
        token=token
    )

    tokenizer = customBPE()
    tokenizer.load(local_path)
    return tokenizer


def load_tokenized_data_from_hf(data_repo_id, token, split="train"):
    from datasets import load_dataset
    try:
        dataset = load_dataset(data_repo_id, split=split, token=token)
        print(f"Extracting tokens for {split} split...")
        
        # No token counting, just load everything fast via an iterator
        def token_generator():
            for row in dataset:
                yield from row["input_ids"]
                
        token_ids = np.fromiter(token_generator(), dtype=np.uint16)
        print(f"Loaded {split} split: {len(token_ids):,} tokens")
        return token_ids
        
    except Exception as e:
        print(f"Could not load '{split}' split from {data_repo_id}: {e}")
        return None


def upload_to_hf(local_path, repo_id, hf_token, path_in_repo):
    api = HfApi(token=hf_token)
    try:
        api.upload_file(
            path_or_fileobj=str(local_path),
            path_in_repo=path_in_repo,
            repo_id=repo_id,
            token=hf_token,
        )
        print(f"Uploaded: {path_in_repo}")
        return True
    except Exception as e:
        print(f"Warning: Failed to upload {path_in_repo}: {e}")
        return False
````

## 10. Load tokenizer and data

````python
print(f"Loading 20k tokenizer for {LANG} from HF...")
tokenizer = load_tokenizer_from_hf(TOKENIZER_REPO_ID, HF_TOKEN)

# Pad vocab_size to exactly hit 25,000,000 parameters!
vocab_size = 20929
print(f"Vocab size (padded for exact 25M params): {vocab_size}")

print(f"Loading tokenized data for {LANG} from HF...")
train_token_ids = load_tokenized_data_from_hf(TOKENIZER_REPO_ID, HF_TOKEN, "train")
val_token_ids = load_tokenized_data_from_hf(TOKENIZER_REPO_ID, HF_TOKEN, "validation")
test_token_ids = load_tokenized_data_from_hf(TOKENIZER_REPO_ID, HF_TOKEN, "test")

if train_token_ids is None:
    raise RuntimeError(
        "Training data failed to load. Do not train on synthetic random data "
        "for the final project run."
    )

# Validation and Test still use PyTorch datasets since they are run sequentially
val_dataset_obj = (
    TextDataset(val_token_ids, SEQ_LEN)
    if val_token_ids is not None
    else None
)

test_dataset_obj = (
    TextDataset(test_token_ids, SEQ_LEN)
    if test_token_ids is not None
    else None
)

val_loader = (
    DataLoader(
        val_dataset_obj,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0
    )
    if val_dataset_obj is not None
    else None
)

test_loader = (
    DataLoader(
        test_dataset_obj,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0
    )
    if test_dataset_obj is not None
    else None
)
````

## 11. Build model, optimizer, and (optionally) resume

````python
model = TransformerLM(
    vocab_size=vocab_size,
    d_model=D_MODEL,
    n_layers=N_LAYERS,
    n_heads=N_HEADS,
    d_ff=D_FF,
    context_length=SEQ_LEN,
    dropout=DROPOUT
).to(device)

total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print(f"Total parameters:     {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LR,
    betas=(0.9, 0.95),
    weight_decay=WEIGHT_DECAY
)


def lr_lambda(current_step):
    if current_step < WARMUP_STEPS:
        return current_step / max(1, WARMUP_STEPS)

    progress = (
        current_step - WARMUP_STEPS
    ) / max(1, MAX_STEPS - WARMUP_STEPS)

    progress = min(max(progress, 0.0), 1.0)

    return 0.5 * (
        1.0 + math.cos(math.pi * progress)
    )


scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer,
    lr_lambda
)

start_step = 0

if RESUME_FROM_HF:
    print("Attempting to resume from latest HF checkpoint...")
    api = HfApi(token=HF_TOKEN)

    try:
        files = api.list_repo_files(
            repo_id=HF_REPO_ID,
            token=HF_TOKEN
        )

        ckpt_files = [
            f for f in files
            if f.startswith("checkpoints/")
            and f.endswith(".pt")
        ]

        step_nums = []

        for f in ckpt_files:
            try:
                step = int(
                    f.split("_")[-1].split(".")[0]
                )
                step_nums.append((step, f))
            except ValueError:
                pass

        if step_nums:
            step_nums.sort(reverse=True)
            latest_ckpt = step_nums[0][1]

            print(f"Downloading checkpoint: {latest_ckpt}")

            ckpt_path = hf_hub_download(
                repo_id=HF_REPO_ID,
                filename=latest_ckpt,
                token=HF_TOKEN
            )

            checkpoint = torch.load(
                ckpt_path,
                map_location=device
            )

            required = {
                "model_state",
                "optimizer_state",
                "scheduler_state",
                "step",
                "config"
            }

            missing = required - set(checkpoint.keys())

            if missing:
                raise RuntimeError(
                    f"Checkpoint is missing required fields: {missing}. "
                    "Use RESUME_FROM_HF=False and start a new run."
                )

            model.load_state_dict(checkpoint["model_state"])
            optimizer.load_state_dict(checkpoint["optimizer_state"])
            scheduler.load_state_dict(checkpoint["scheduler_state"])

            start_step = checkpoint["step"] + 1

            print(f"Resuming from step {start_step}")

        else:
            print("No checkpoints found - starting from scratch.")

    except Exception as e:
        raise RuntimeError(
            f"Could not safely resume from HF: {e}"
        )

# Verify the architecture before spending compute.
x_check = torch.randint(0, vocab_size, (1, SEQ_LEN)).to(device)
verify_causal_mask(model, x_check, vocab_size)
````

## 12. Training loop

````python
config_dict = {
    "lang": LANG,
    "vocab_size": vocab_size,
    "d_model": D_MODEL,
    "n_layers": N_LAYERS,
    "n_heads": N_HEADS,
    "d_ff": D_FF,
    "context_length": SEQ_LEN,
    "dropout": DROPOUT,
    "weight_tying": True,
    "batch_size": BATCH_SIZE,
    "learning_rate": LR,
    "weight_decay": WEIGHT_DECAY,
    "warmup_steps": WARMUP_STEPS,
    "max_steps": MAX_STEPS,
    "target_train_tokens": TARGET_TRAIN_TOKENS,
    "trainable_parameters": trainable_params,
    "total_parameters": total_params,
    "merges": MERGES
}

wandb.init(
    project=WANDB_PROJECT,
    entity=WANDB_ENTITY,
    name=f"lma-{LANG}-25M",
    config=config_dict
)
````

````python
global_step = start_step
pbar = tqdm(
    total=MAX_STEPS,
    desc="Training",
    initial=start_step,
    unit="step"
)

model.train()
last_loss = 0.0

while global_step < MAX_STEPS:

    # Highly optimized numpy batching entirely bypassing Python data structures in tight loop
    ix = torch.randint(0, len(train_token_ids) - SEQ_LEN - 1, (BATCH_SIZE,)).tolist()
    x = torch.tensor(np.stack([train_token_ids[i : i + SEQ_LEN] for i in ix]).astype(np.int64), dtype=torch.long, device=device)
    y = torch.tensor(np.stack([train_token_ids[i + 1 : i + 1 + SEQ_LEN] for i in ix]).astype(np.int64), dtype=torch.long, device=device)

    logits, loss = model(x, y)
    last_loss = loss.item()

    optimizer.zero_grad(set_to_none=True)
    loss.backward()

    # Prevent occasional exploding updates.
    torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)

    optimizer.step()
    scheduler.step()

    if global_step % LOG_STEPS == 0:
        wandb.log({
            "train/loss": last_loss,
            "train/perplexity": math.exp(last_loss),
            "train/lr": optimizer.param_groups[0]["lr"],
            "train/step": global_step,
        })

    pbar.update(1)

    if global_step % 100 == 0:
        pbar.set_postfix({
            "loss": f"{last_loss:.4f}",
            "ppl": f"{math.exp(min(last_loss, 20)):.2f}",
            "lr": f"{optimizer.param_groups[0]['lr']:.2e}"
        })

    if global_step > 0 and global_step % CHECKPOINT_STEPS == 0:
        ckpt_name = f"checkpoint_step_{global_step}.pt"
        ckpt_path = f"/kaggle/working/{ckpt_name}"

        torch.save({
            "step": global_step,
            "model_state": model.state_dict(),
            "optimizer_state": optimizer.state_dict(),
            "scheduler_state": scheduler.state_dict(),
            "loss": last_loss,
            "config": config_dict,
        }, ckpt_path)

        upload_to_hf(
            ckpt_path,
            HF_REPO_ID,
            HF_TOKEN,
            f"checkpoints/{ckpt_name}"
        )

        wandb.log({
            "checkpoint_step": global_step
        })

        print(f"\nSaved checkpoint at step {global_step}")

    if (
        global_step > 0
        and global_step % EVAL_STEPS == 0
        and val_loader is not None
    ):
        model.eval()

        ppl = compute_perplexity(
            model,
            val_loader,
            device,
            max_batches=200
        )

        val_loss = math.log(ppl)

        wandb.log({
            "val/loss": val_loss,
            "val/perplexity": ppl,
            "val/step": global_step,
        })

        print(
            f"\nStep {global_step}: "
            f"Val Loss = {val_loss:.4f} | "
            f"Val Perplexity = {ppl:.2f}"
        )

        model.train()

    global_step += 1

pbar.close()

print(f"\nTraining completed, step {global_step}")
print(f"Approx. tokens processed: {global_step * TOKENS_PER_STEP:,}")
````
